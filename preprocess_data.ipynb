```python
import tarfile
import pandas as pd

# Open archive
with tarfile.open('raw_data/ThirtyMusic.tar.gz', 'r:gz') as tar:
    # Extract all files
    tar.extractall('temp_extract/')

# Now read each file
print("=== ENTITIES ===")

# Users
print("\n1. Users:")
try:
    users_df = pd.read_csv('temp_extract/entities/users.idomaar', sep='\t', header=None)
    print(f"Size: {users_df.shape}")
    print("First 3 records:")
    print(users_df.head(3))
except Exception as e:
    print(f"Error reading users: {e}")

# Playlists
print("\n2. Playlists:")
try:
    playlists_df = pd.read_csv('temp_extract/entities/playlist.idomaar', sep='\t', header=None)
    print(f"Size: {playlists_df.shape}")
    print("First 3 records:")
    print(playlists_df.head(3))
except Exception as e:
    print(f"Error reading playlists: {e}")

# Tracks
print("\n3. Tracks:")
try:
    tracks_df = pd.read_csv('temp_extract/entities/tracks.idomaar', sep='\t', header=None)
    print(f"Size: {tracks_df.shape}")
    print("First 3 records:")
    print(tracks_df.head(3))
except Exception as e:
    print(f"Error reading tracks: {e}")

# Tags
print("\n4. Tags:")
try:
    tags_df = pd.read_csv('temp_extract/entities/tags.idomaar', sep='\t', header=None)
    print(f"Size: {tags_df.shape}")
    print("First 3 records:")
    print(tags_df.head(3))
except Exception as e:
    print(f"Error reading tags: {e}")

# Persons
print("\n5. Persons:")
try:
    persons_df = pd.read_csv('temp_extract/entities/persons.idomaar', sep='\t', header=None)
    print(f"Size: {persons_df.shape}")
    print("First 3 records:")
    print(persons_df.head(3))
except Exception as e:
    print(f"Error reading persons: {e}")

# Albums
print("\n6. Albums:")
try:
    albums_df = pd.read_csv('temp_extract/entities/albums.idomaar', sep='\t', header=None)
    print(f"Size: {albums_df.shape}")
    print("First 3 records:")
    print(albums_df.head(3))
except Exception as e:
    print(f"Error reading albums: {e}")

print("\n=== RELATIONS ===")

# Events
print("\n7. Events:")
try:
    events_df = pd.read_csv('temp_extract/relations/events.idomaar', sep='\t', header=None)
    print(f"Size: {events_df.shape}")
    print("First 3 records:")
    print(events_df.head(3))
except Exception as e:
    print(f"Error reading events: {e}")

# Sessions
print("\n8. Sessions:")
try:
    sessions_df = pd.read_csv('temp_extract/relations/sessions.idomaar', sep='\t', header=None)
    print(f"Size: {sessions_df.shape}")
    print("First 3 records:")
    print(sessions_df.head(3))
except Exception as e:
    print(f"Error reading sessions: {e}")

# Favorite tracks
print("\n9. Favorite tracks:")
try:
    love_df = pd.read_csv('temp_extract/relations/love.idomaar', sep='\t', header=None)
    print(f"Size: {love_df.shape}")
    print("First 3 records:")
    print(love_df.head(3))
except Exception as e:
    print(f"Error reading favorite tracks: {e}")
```

```python
import pandas as pd
import numpy as np

# Read data
df = pd.read_parquet('../data/listens.parquet')

# Create empty list for result
result_rows = []

# Iterate over each user
for _, row in df.iterrows():
    uid = row['uid']
    timestamps = row['timestamp']
    item_ids = row['item_id']
    
    # Create records for each user interaction
    for timestamp, item_id in zip(timestamps, item_ids):
        result_rows.append({
            'userid': uid,
            'itemid': item_id,
            'timestamp': timestamp
        })

# Create new DataFrame
processed_df = pd.DataFrame(result_rows)

```

```python
processed_df['userid'].nunique()
```

```python
processed_df = processed_df.sort_values(['userid', 'timestamp'])
```

```python
processed_df.userid = processed_df.userid.astype('category').cat.codes
processed_df.itemid = processed_df.itemid.astype('category').cat.codes + 1 
```

```python
ds_name = 'yanbda-50m'
processed_df.to_csv(f'preprocessed_data/{ds_name}.csv', index=False)
```
